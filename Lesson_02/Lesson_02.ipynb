```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Rainfall": [10, 20, 30, 40, 50],
    "Soil_Moisture": [15, 25, 20, 60, 45]
})
```

```python
x1 =df["Rainfall"].to_numpy()
x2 =df["Soil_Moisture"].to_numpy()
```

```python
x1 = np.array(x1)
x2 = np.array(x2)
```

```python
x1_sorted_index = np.argsort(x1)
x2_sorted_index = np.argsort(x2)

print(x1_sorted_index)
print(x2_sorted_index)
```

```python
rank_x1 = np.empty_like(x1_sorted_index)
rank_x2 = np.empty_like(x2_sorted_index)

print(rank_x1)
print(rank_x2)
```

```python
rank_x1[x1_sorted_index] = np.arange(1, (len(x1_sorted_index) + 1))
rank_x2[x2_sorted_index] = np.arange(1, (len(x2_sorted_index) + 1))

print(rank_x1)
print(rank_x2)
```

```python
rank_difference = rank_x1 - rank_x2

print(rank_difference)
```

```python
print(np.sum(rank_difference ** 2) * 6)
print(rank_difference.shape[0] * (rank_difference.shape[0] ** 2 - 1))
```

```python
spearmans_correlation = 1 - (np.sum(rank_difference ** 2) * 6) / (rank_difference.shape[0] * (rank_difference.shape[0] ** 2 - 1))

print(spearmans_correlation)
```

```python
# Calculate Automatically and This Method is Correct For Both Tied and Untied Ranks
```

```python
spearman_r = np.corrcoef(rank_x1, rank_x2)[0, 1]

print(spearman_r)
```

```python
from scipy.stats import spearmanr

result = spearmanr(x1, x2)

print("Spearman correlation:", result.statistic)
print("P-value:", result.pvalue)
```

```python
# The P value is greater than 0.05 that means we will accept the null hyptohesis and the null hypothesis is there is no significant relationship between rainfall and soil moisture
```

```python
# Rank Calculation
def tied_ranks(x):
    order_x = np.argsort(x)
    rank_x = np.empty_like(order_x, dtype= float)
    rank_x[order_x] = np.arange(1, (len(x)+1))
    unique_values, counts = np.unique(x, return_counts= True)
    if np.any(counts > 1):
        for value, count in zip(unique_values, counts):
            if count > 1:
                positions= np.where(x == value)[0]
                rank_x[positions] = np.mean(rank_x[positions])
    else:
        print("No Duplicates Were Found")
    return rank_x
```

```python
rank_rainfall = tied_ranks(rainfall)
rank_rd = tied_ranks(river_discharge)
rank_ele = tied_ranks(elevation)
rank_temp = tied_ranks(temperature)
```

```python
print(rank_rainfall)
print(rank_rd)
print(rank_ele)
print(rank_temp)
```

```python
spearman_r_for_rainfall_rd = np.corrcoef(rank_rainfall, rank_rd)[0,1]

print(spearman_r_for_rainfall_rd)
```

```python
result_1 = spearmanr(rank_rainfall, rank_rd)

print("Spearman correlation:", result_1.statistic)
print("P-value:", result_1.pvalue)
```

```python
rank_df = pd.DataFrame(np.column_stack((rank_rainfall, rank_rd, rank_ele, rank_temp)), columns= ["rank_rainfall", "rank_rd", "rank_ele", "rank_temp"])
```

```python
spearman_corr_all = np.corrcoef(rank_df, rowvar= False)

print(spearman_corr_all)
```

```python
result_2 = spearmanr(rank_df)

print("Spearman correlation:", result_2.statistic)
print("P-value:", result_2.pvalue)
```
