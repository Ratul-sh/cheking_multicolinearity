```python
# Spearman's Rank Correlation
```

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Rainfall": [10, 20, 30, 40, 50],
    "Soil_Moisture": [15, 25, 20, 60, 45]
})
```

```python
x1 =df["Rainfall"].to_numpy()
x2 =df["Soil_Moisture"].to_numpy()
```

```python
x1 = np.array(x1)
x2 = np.array(x2)
```

```python
x1_sorted_index = np.argsort(x1)
x2_sorted_index = np.argsort(x2)

print(x1_sorted_index)
print(x2_sorted_index)
```

```python
rank_x1 = np.empty_like(x1_sorted_index)
rank_x2 = np.empty_like(x2_sorted_index)

print(rank_x1)
print(rank_x2)
```

```python
rank_x1[x1_sorted_index] = np.arange(1, (len(x1_sorted_index) + 1))
rank_x2[x2_sorted_index] = np.arange(1, (len(x2_sorted_index) + 1))

print(rank_x1)
print(rank_x2)
```

```python
rank_difference = rank_x1 - rank_x2

print(rank_difference)
```

```python
print(np.sum(rank_difference ** 2) * 6)
print(rank_difference.shape[0] * (rank_difference.shape[0] ** 2 - 1))
```

```python
spearmans_correlation = 1 - (np.sum(rank_difference ** 2) * 6) / (rank_difference.shape[0] * (rank_difference.shape[0] ** 2 - 1))

print(spearmans_correlation)
```

```python
# Calculate Automatically and This Method is Correct For Both Tied and Untied Ranks
```

```python
spearman_r = np.corrcoef(rank_x1, rank_x2)[0, 1]

print(spearman_r)
```

```python
from scipy.stats import spearmanr

result = spearmanr(x1, x2)

print("Spearman correlation:", result.statistic)
print("P-value:", result.pvalue)
```

```python
# The P value is greater than 0.05 that means we will accept the null hyptohesis and the null hypothesis is there is no significant relationship between rainfall and soil moisture
```
