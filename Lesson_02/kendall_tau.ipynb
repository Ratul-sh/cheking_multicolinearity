```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Rainfall": [10, 20, 30, 40],
    "River_Discharge": [20, 40, 30, 60]
})

x = df["Rainfall"].to_numpy()
y = df["River_Discharge"].to_numpy()
```

```python
# Calculating number of variable pairs
# number of pairs = p*(p-1) / 2
p = df.shape[1] # Number of variables in the dataset
varible_pairs = (p*(p-1)) // 2
varible_pairs
```

```python
# Calculating all possible observation pairs
n = len(x)
pairs = (n*(n-1)) // 2
pairs
```

```python
# Calculating pair type
def kendall_tau(x, y):
    n = len(x)
    pairs = (n*(n-1)) // 2
    concordant = 0
    discordant = 0
    ties_x = 0
    ties_y = 0
    joint_ties = 0

    if len(x) != len(y):
        raise ValueError("Both arrays must have equal lengths")
    else:
        for i in range(len(x)-1):
            for j in range(i, len(y)-1):
                pair_x = x[i] - x[j+1]
                pair_y = y[i] - y[j+1]
                pair_couple = pair_x * pair_y
                if pair_couple > 0:
                    concordant += 1
                elif pair_couple < 0:
                    discordant += 1
                elif pair_x == 0 and pair_y == 0:
                    joint_ties += 1
                elif pair_x == 0:
                    ties_x += 1
                else:
                    ties_y += 1
    if ties_x > 0 or ties_y > 0 or joint_ties > 0:
        kendall_tau_b = np.subtract(concordant, discordant) / np.sqrt((concordant + discordant + ties_x) * (concordant + discordant + ties_y))
        return f"Ties founded, therefore calculated Kendall Tau-b: {round(kendall_tau_b, 3)}"
    else:
        kendall_tau_a = np.subtract(concordant, discordant) / (pairs)
        return f"Ties not founded, therefore calculated Kendall Tau-a: {round(kendall_tau_a, 3)}"
        
```
