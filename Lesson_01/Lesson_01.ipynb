```python
import numpy as np
import pandas as pd
data = {
    "Rainfall": [12, 18, 25, 32, 40, 48],
    "Soil_Moisture": [22, 27, 35, 42, 49, 58],
    "Elevation": [100, 95, 88, 80, 75, 65],
    "Temperature": [30, 32, 31, 35, 34, 38]
}

df = pd.DataFrame(data)
```

```python
# Calculating Mean
rainfall_mean = df["Rainfall"].mean()
soil_moisture_mean = df["Soil_Moisture"].mean()
print(rainfall_mean)
print(soil_moisture_mean)
```

```python
# Calculating Deviation

df["rainfall_deviation"] = df["Rainfall"] - rainfall_mean
df["sm_deviation"] = df["Soil_Moisture"] - soil_moisture_mean

df
```

```python
# Calculating Sample Covariance
sample_cov = (df["rainfall_deviation"].multiply(df["sm_deviation"]).sum()) / (len(df) - 1)

print(sample_cov)
```

```python
# Calculating Standard Deviation
std_rainfall = (df["rainfall_deviation"].pow(2).sum() / (len(df) - 1)) ** 0.5
std_sm = (df["sm_deviation"].pow(2).sum() / (len(df) - 1)) ** 0.5

print(std_rainfall)
print(std_sm)
```

```python
# Calculating Pearson Correlation
corr = sample_cov / (std_rainfall * std_sm)

print(corr)
```

```python
# Converting Data Frame to Numpy

x = np.array(df)
print(x)
```

```python
x =  np.delete(x, [4, 5], axis = 1)

print(x)
```

```python
means = np.mean(x, axis = 0)

print(means)
```

```python
x_deviated = x - means

print(x_deviated)
```

```python
x_deviated_transpose = x_deviated.T

print(x_deviated_transpose)
```

```python
sample_cov_in_numpy = (
    x_deviated_transpose @ x_deviated / (x.shape[0] -1)
)

print(sample_cov)
```

```python
std = np.sqrt(np.diag(sample_cov_in_numpy))

print(std)
```

```python
sample_corr = (
    sample_cov_in_numpy / np.outer(std,std)
)

print(sample_corr)
```
