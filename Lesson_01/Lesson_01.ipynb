```python
import numpy as np
import pandas as pd
data = {
    "Rainfall": [12, 18, 25, 32, 40, 48],
    "Soil_Moisture": [22, 27, 35, 42, 49, 58],
    "Elevation": [100, 95, 88, 80, 75, 65],
    "Temperature": [30, 32, 31, 35, 34, 38]
}

df = pd.DataFrame(data)
```

```python
# Calculating Mean
rainfall_mean = df["Rainfall"].mean()
soil_moisture_mean = df["Soil_Moisture"].mean()
print(rainfall_mean)
print(soil_moisture_mean)
```

```python
# Calculating Deviation

df["rainfall_deviation"] = df["Rainfall"] - rainfall_mean
df["sm_deviation"] = df["Soil_Moisture"] - soil_moisture_mean

df
```

```python
# Calculating Sample Covariance
sample_cov = (df["rainfall_deviation"].multiply(df["sm_deviation"]).sum()) / (len(df) - 1)

print(sample_cov)
```

```python
# Calculating Standard Deviation
std_rainfall = (df["rainfall_deviation"].pow(2).sum() / (len(df) - 1)) ** 0.5
std_sm = (df["sm_deviation"].pow(2).sum() / (len(df) - 1)) ** 0.5

print(std_rainfall)
print(std_sm)
```

```python
# Calculating Pearson Correlation
corr = sample_cov / (std_rainfall * std_sm)

print(corr)
```

```python
sample_cov_in_numpy = (
    x_deviated_transpose @ x_deviated / (x.shape[0] -1)
)

print(sample_cov)
```

```python
std = np.sqrt(np.diag(sample_cov_in_numpy))

print(std)
```

```python
sample_corr = (
    sample_cov_in_numpy / np.outer(std,std)
)

print(sample_corr)
```

```python
data_new = {
    "Rainfall": [15, 22, 30, 35, 42, 50, 58, 65],
    "Soil_Moisture": [25, 30, 38, 40, 48, 52, 60, 68],
    "Elevation": [100, 95, 90, 85, 80, 75, 70, 65],
    "Temperature": [28, 29, 30, 31, 33, 34, 35, 36]
}

df_new = pd.DataFrame(data_new)
```

```python
array = np.array(df_new)

print(array)
```

```python
means_new = np.mean(array, axis = 0)

print(means_new)
```

```python
deviated_array = array - means_new

print(deviated_array)
```

```python
transpose_deviated_array = deviated_array.T

print(transpose_deviated_array)
```

```python
new_sample_cov = transpose_deviated_array @ deviated_array / (array.shape[0] - 1)

print(new_sample_cov)
```

```python
new_std = np.sqrt(np.diag(new_sample_cov))

print(new_std)
```

```python
pearson_correlation = new_sample_cov / np.outer(new_std, new_std)

print(pearson_correlation)
```

```python
print(f"Pearson Correlation Between Rainfall and Soil Moisture: {pearson_correlation[0, 1]}")
```
